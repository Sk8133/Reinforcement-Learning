```python
# Assignment 8 - Reinforcement Learning
# Calculate State Value and Action Value
# Without using Bellman Equation

# Discount factor
gamma = 0.9

# Grid World transitions
# State : {Action : (Next State, Reward)}
grid = {
    "S0": {
        "right": ("S1", 1),
        "up": ("S0", -1)
    },
    "S1": {
        "down": ("S3", 1)
    },
    "S3": {
        "right": ("S4", 1)
    },
    "S4": {
        "up": ("Terminal", 10)
    }
}

# Given Policy
policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}


def calculate_return(start_state):
    """Calculate discounted return by following the policy."""

    state = start_state
    total_return = 0
    discount = 1

    while state != "Terminal":
        action = policy[state]
        next_state, reward = grid[state][action]

        total_return += discount * reward
        discount *= gamma

        state = next_state

    return total_return


def state_value_function(state):
    return calculate_return(state)


def action_value_function(state, action):
    next_state, reward = grid[state][action]

    if next_state == "Terminal":
        return reward

    return reward + gamma * calculate_return(next_state)


# State Value for S0
v_s0 = state_value_function("S0")

# Action Value for S0 and UP
q_s0_up = action_value_function("S0", "up")

print("State Value V(S0) =", v_s0)
print("Action Value Q(S0, UP) =", q_s0_up)
```

```python
import random

# -----------------------------------
# Part A - Machine Initialization
# -----------------------------------

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# N = Number of times each machine is selected
N = {
    'A': 0,
    'B': 0,
    'C': 0
}

# R = Total reward received from each machine
R = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Q = Estimated average reward of each machine
Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

total_reward = 0


# -----------------------------------
# Part B - Initial Test
# -----------------------------------

print("INITIAL TEST FOR ALL MACHINES")

for machine in machines:

    reward = float(input(f"Enter initial reward for Machine {machine}: "))

    N[machine] = 1
    R[machine] = reward
    Q[machine] = reward

print("\nInitial Tables")

print("Machine\tN\tR\tQ")

for machine in machines:
    print(machine, "\t", N[machine], "\t", R[machine], "\t", Q[machine])


# -----------------------------------
# Part C, D and E
# Modified Epsilon-Greedy Strategy
# -----------------------------------

print("\nSTARTING EPSILON-GREEDY PLAYS")

for play in range(1, N_play + 1):

    print("\n-------------------------")
    print("Play Number:", play)

    # Find highest Q-value
    best_machine = max(machines, key=lambda x: Q[x])

    # Generate random number
    random_number = random.random()

    print("Random number =", round(random_number, 3))

    # Exploration
    if random_number < epsilon:

        print("Exploration")

        # Select machines except the best machine
        other_machines = [m for m in machines if m != best_machine]

        selected_machine = random.choice(other_machines)

    # Exploitation
    else:

        print("Exploitation")

        selected_machine = best_machine

    print("Play with Machine", selected_machine)

    # -----------------------------------
    # Part D - Reward Collection
    # -----------------------------------

    reward = float(input("Enter reward: "))

    # -----------------------------------
    # Part E - Q-Value Update
    # -----------------------------------

    # Increment count
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Incremental Sample Average Formula:
    #
    # Q_new = Q_old + (1 / N) * (Reward - Q_old)

    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    # Update total player reward
    total_reward += reward

    print("\nUpdated Values")

    print("Machine =", selected_machine)
    print("N =", N[selected_machine])
    print("R =", R[selected_machine])
    print("Q =", round(Q[selected_machine], 2))


# -----------------------------------
# Part F - Display Final Results
# -----------------------------------

print("\n=================================")
print("FINAL RESULTS")
print("=================================")

print("\nMachine\tParticipation(N)\tTotal Reward(R)\tQ-value")

for machine in machines:

    print(
        machine,
        "\t",
        N[machine],
        "\t\t\t",
        R[machine],
        "\t\t",
        round(Q[machine], 2)
    )

print("\nTotal Reward Received by Player =", total_reward)

print("\nBest Machine Based on Final Q-value =",
      max(machines, key=lambda x: Q[x]))
```
